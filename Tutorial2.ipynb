```python
#Q1

import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is numpy 2 by 2 array")
print(array_2x2)
```

```python
#Q2

random3x4 = np.random.rand(3,4)

print("Random 3 by 4 array: ")
print(random3x4)
```

```python
#Q3

ones = np.ones((3,3))
zeros = np.zeros((3,3))

print("Ones:")
print(ones)

print("Zeros:")
print(zeros)
```

```python
#Q4

arr_6x3 = np.random.rand(6,3)

print("Original array: ")
print(arr_6x3)

print("Transposed array:")
print(arr_6x3.T)
```

```python
#Q5

arr_3x2 = np.random.rand(3,2)
arr_2x4 = np.random.rand(2,4)

print("Array 1:")
print(arr_3x2)

print("Array 2:")
print(arr_2x4)

print("Dot product:")
print(arr_3x2.dot(arr_2x4))
```
